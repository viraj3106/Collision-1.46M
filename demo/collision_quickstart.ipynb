## 2. Natural Web Grounded Answering (ChatGPT / Gemini Style)

Ask open-domain questions and receive naturally synthesized, fact-grounded responses with verified citations.

```python
from collision import CollisionService

service = CollisionService()

# Query 1: Technology & Release Updates
res1 = service.ask("What is the latest release version of PyTorch in 2025?", mode="WEB")
print("=" * 60)
print("QUERY: Latest PyTorch Release")
print("-" * 60)
print(res1["answer"])
print("\nSources:", [s["title"] for s in res1.get("sources", [])])

# Query 2: Hardware Specifications
res2 = service.ask("What are the core specifications of the Apple M4 chip?", mode="WEB")
print("\n" + "=" * 60)
print("QUERY: Apple M4 Specifications")
print("-" * 60)
print(res2["answer"])

# Query 3: Origin & Biographies
res3 = service.ask("Who created Python and when was it first released?", mode="WEB")
print("\n" + "=" * 60)
print("QUERY: Python History")
print("-" * 60)
print(res3["answer"])
```

## 3. Industrial NLP Suite (`collision.nlp`)

Run zero-latency NLP tasks without any external heavy models.

```python
from collision.nlp import CollisionNLPEngine

sample_text = """
Quantum computing leverages superposition and entanglement to execute complex quantum algorithms.
The engineering research team achieved unprecedented fault-tolerant gate fidelity on experimental hardware.
"""

# 1. TextRank Keyphrase Extraction
kp = CollisionNLPEngine.extract_keywords(sample_text)
print("🏷️ Keyphrases:", kp.keyphrases)

# 2. Multi-Domain Topic Classification (10 Domains)
topic = CollisionNLPEngine.classify_topic(sample_text)
print(f"📊 Topic: {topic.primary_topic} (Confidence: {topic.confidence*100:.0f}%)")

# 3. Tone & Formality Analysis
tone = CollisionNLPEngine.analyze_tone(sample_text)
print(f"🎭 Tone: {tone.primary_tone} | Formality: {tone.formality_score}/100")

# 4. Grammar & Typographical Proofreader
proof = CollisionNLPEngine.proofread("I ate a apple on the the kitchen table .")
print(f"✍️ Proofread: '{proof.corrected_text}'")

# 5. Readability & Complexity Indices
readability = CollisionNLPEngine.analyze_readability(sample_text)
print(f"📈 Flesch Reading Ease: {readability.flesch_reading_ease:.1f} ({readability.reading_level})")
```

## 4. Deterministic Problem Solving (Math, Geometry, Stats & Units)

COLLISION includes deterministic calculation engines for 100% exact solutions.

```python
math_queries = [
    "What is 45 * 12 + 180 / 4?",
    "Calculate the hypotenuse of a right triangle with legs 9 and 12.",
    "Convert 100 kilometers to miles.",
    "What is the standard deviation of 12, 18, 24, 30, 36?"
]

for q in math_queries:
    res = service.ask(q, mode="AUTO")
    print(f"Q: {q}\nA: {res['answer']}\n")
```

## 5. Raw Neural Model Text Generation

Generate text directly from the 10.28M parameter base causal transformer.

```python
from collision.answering.engine import CollisionAnsweringEngine

engine = CollisionAnsweringEngine()
result = engine.answer(
    question="The future of artificial intelligence",
    max_tokens=40,
    temperature=0.7
)
print(f"Generated: {result.text}")
print(f"Speed: {result.tokens_per_second:.1f} tokens/sec on CPU")
```
